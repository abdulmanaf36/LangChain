```python
from flashrank.Ranker import Ranker, RerankRequest
```

```python
documents = [
    "This a list which containing sample documents.",
    "Keywords are important for keyword-based search.",
    "Document analysis involves extracting keywords.",
    "Keyword-based search relies on sparse embeddings.",
    "Understsnding document structure aids in keyword extraction.",
    "Efficient keyword extraction enhances search accuracy.",
    "Semacntic similarity improves document retrieval performance.",
    "Machine learning techniques can optimize keyword extraction methods."
]
len(documents)

query = "Natural language processing techniques enhances keyword extraction efficiency."
```

```python
documents = [{"text": doc, "id": i, "meta": {}} for i, doc in enumerate(documents)]
```

```python
ranker = Ranker(model_name="rank-T5-flan", cache_dir="/cluster/home/abdulman/NTNU_Exchange/RAG_Chatbot_FastAPI/opt")
```

```python
rerankrequest = RerankRequest(query=query, passages=documents)
results = ranker.rerank(rerankrequest)
```

```python
for result in results[:4]:
    print(f"Document: {result['text']}, Score: {result['score']}")
```
